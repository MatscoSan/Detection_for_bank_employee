# Customer class

```python
class Customer :
    def __init__(self, id, parent = None, role = None):
        self.id = id
        self.parent = parent
        self.role = role
        if (role == 'M') :
            self.role = "Member"
        if (role == 'VP') :
            self.role = "Viewing Party"
        if (role == 'B') :
            self.role = "Business"
```

# Create df and imports

```python
import pandas as pd

df = pd.read_csv('data/partner_role.csv')
display(df)
```

## Relations

```python

df1 = df.drop(columns=["relationship_start_date", "relationship_end_date", "partner_class_code"])
customers_relationship = []

for index, row in df1.iterrows():
    if row['entity_type'] == "BR": 
        customer = Customer(row['partner_id'], row['associated_partner_id'], row['br_type_code'])  
        customers_relationship.append(customer)

```

```python
import json, os
# Exporte la liste customers_relationship dans un fichier JS utilisable côté front sans forcer la lecture des CSV
nodes = []
# Essaie de charger les noms depuis data/partner.csv si disponible
id2name = {}
try:
    import pandas as _pd
    if os.path.exists('data/partner.csv'):
        p = _pd.read_csv('data/partner.csv', dtype={'partner_id': str})
        id2name = dict(zip(p['partner_id'].astype(str), p['partner_name'].astype(str)))
except Exception as e:
    # pandas peut ne pas être disponible ou le fichier absent; on continue sans noms
    id2name = {}
for c in customers_relationship:
    cid = str(c.id)
    pid = None if c.parent is None else str(c.parent)
    name = id2name.get(cid, cid)
    parent_name = id2name.get(pid, None) if pid is not None else None
    nodes.append({'id': cid, 'name': name, 'parent': pid, 'parent_name': parent_name, 'role': c.role})
out_path = 'data/customers_relationship_data.js'
os.makedirs(os.path.dirname(out_path), exist_ok=True)
with open(out_path, 'w', encoding='utf-8') as f:
    f.write('window.CUSTOMERS_RELATIONSHIP = ')
    json.dump({'nodes': nodes}, f, ensure_ascii=False, indent=2)
    f.write(';')
print(f'Wrote {len(nodes)} nodes to {out_path}')
# affiche un aperçu
try:
    import pandas as pd
    display(pd.DataFrame(nodes).head())
except Exception:
    print(nodes[:5])
```
