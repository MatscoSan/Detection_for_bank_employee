# Customer class

```python
class Customer :
    def __init__(self, id, parent = None, role = None):
        self.id = id
        self.parent = parent
        self.role = role
        if (role == 'M') :
            self.role = "Member"
        if (role == 'VP') :
            self.role = "Viewing Party"
        if (role == 'B') :
            self.role = "Business"
```

# Create df and imports

```python
import pandas as pd

df = pd.read_csv('data/partner_role.csv')
display(df)
```

## Relations

```python

df1 = df.drop(columns=["relationship_start_date", "relationship_end_date", "partner_class_code"])
customers_relationship = []

for index, row in df1.iterrows():
    if row['entity_type'] == "BR": 
        customer = Customer(row['partner_id'], row['associated_partner_id'], row['br_type_code'])  
        customers_relationship.append(customer)
```
